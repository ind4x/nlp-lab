```python
# for dataframe analysis and manipulation.
import pandas as pd
import numpy as np
import regex as re

# for data preprocessing
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer

# for model Building
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# for removing stopwords
from nltk.corpus import stopwords
stop = stopwords.words('english')

#!python -m nltk.downloader stopwords
```

# Loading the dataset

```python
df = pd.read_table('SpamCollection.txt', names = ['Label', 'Message'])
df.head()
```

## About the dataset

- The dataset contains labeled emaols, where each mail is classified as spam. It has both the email text and corresponding labels

```python
df.info()
```

## Analysis 

- There are no messing value in the dataset
- The target value is in categorical form

# Data Preparation

## Incoding target variable

```python
le = LabelEncoder()
df['Label'] = le.fit_transform(df['Label'])
```

```python
df.head()
```

## Data Cleaning

```python
def data_cleaning(text):
    text = str(text).lower()
    text = re.sub('[^a-z]', ' ',  text)
    words = text.split()
    imp_words = [word for word in words if word not in stop]
    return ' '.join(imp_words)
```

```python
df['Message'] = df['Message'].apply(lambda x: data_cleaning(x))
# lambda est une fct anonime qui permet d'appliquer une fonction à chaque élément d'une série ou d'une colonne d'un DataFrame. Ici, elle est utilisée pour appliquer la fonction data_cleaning à chaque message de la colonne 'Message' du DataFrame df.
```

```python
df.head()
```

## Train test split

```python
X = df['Message']
y = df['Label']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
```

## Feature extraction

- Converting the text data into numerical features using countvectoriser

```python
df.Message.head()
```

```python
cv = CountVectorizer()
# CountVectorizer est une classe de la bibliothèque scikit-learn qui convertit une collection de documents textuels en une matrice de caractéristiques numériques. Chaque ligne de la matrice représente un document, et chaque colonne représente un mot unique dans l'ensemble des documents. La valeur dans chaque cellule indique le nombre d'occurrences du mot correspondant dans le document correspondant.

X_train = cv.fit_transform(X_train)
X_test = cv.transform(X_test)

# fit_transform() est utilisé pour ajuster le CountVectorizer aux données d'entraînement et transformer ces données en une matrice de caractéristiques. transform() est utilisé pour transformer les données de test en utilisant le même CountVectorizer ajusté sur les données d'entraînement.
```

```python
Indexes = pd.DataFrame(X_train.toarray(), columns=cv.get_feature_names_out())
print(cv.vocabulary_)
```

```python
occurences = pd.DataFrame(X_train.toarray(), columns=cv.get_feature_names_out())
print(occurences)
```

# Model Training

```python
# Model Training
multi_nb = MultinomialNB()
multi_nb.fit(X_train, y_train)
```

MultinomialNB()

```python
# Model Evaluation
y_pred = multi_nb.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred)

# Print scores
print("Accuracy:", round(accuracy, 4))
print("\nConfusion Matrix:\n", conf_matrix)
print("\nClassification Report:\n", report)
```
